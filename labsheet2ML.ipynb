```python
# Q1. Load a dataset and identify missing values in each column.
import pandas as pd

df = pd.read_csv("original_dataset.csv")

print(df.isnull().sum())
```

```python
# Q2. Display the percentage of missing values in every feature.
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
#Q3. Remove rows containing missing values from the dataset.
df_no_missing_rows = df.dropna()

print(df_no_missing_rows)
```

```python
# Q4. Remove columns having more than 50% missing values.
missing_percentage = df.isnull().mean() * 100

df_removed_columns = df.drop(
    columns=missing_percentage[missing_percentage > 50].index
)

print(df_removed_columns)
```

```python
# Q4. Remove columns having more than 50% missing values.
missing_percentage = df.isnull().mean() * 100

df_removed_columns = df.drop(
    columns=missing_percentage[missing_percentage > 50].index
)

print(df_removed_columns)
```

```python
# Q5. Replace missing numerical values using the mean.
import numpy as np
df_mean = df.copy()

numeric_columns = df_mean.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:
    df_mean[col] = df_mean[col].fillna(
        df_mean[col].mean()
    )

print(df_mean)
```

```python
# Q6. Replace missing numerical values using the median.
df_median = df.copy()

numeric_columns = df_median.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:
    df_median[col] = df_median[col].fillna(
        df_median[col].median()
    )

print(df_median)
```

```python
# Q7. Replace missing categorical values using the mode.
df_mode = df.copy()

categorical_columns = df_mode.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    df_mode[col] = df_mode[col].fillna(
        df_mode[col].mode()[0]
    )

print(df_mode) 
```

```python
# Q8. Fill missing values using Forward Fill.
df_ffill = df.copy()

df_ffill = df_ffill.ffill()

print(df_ffill)
```

```python
# Q9. Fill missing values using Backward Fill.
df_bfill = df.copy()

df_bfill = df_bfill.bfill()

print(df_bfill)
```

```python
# Q10. Compare the dataset before and after handling missing values.
print("Missing values before treatment:")
print(df.isnull().sum())

df_clean = df.copy()

numeric_columns = df_clean.select_dtypes(
    include=np.number
).columns

categorical_columns = df_clean.select_dtypes(
    include="object"
).columns

for col in numeric_columns:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].median()
    )

for col in categorical_columns:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].mode()[0]
    )
print("\nMissing values after treatment:")
print(df_clean.isnull().sum())
```

```python
# Q11. Detect outliers using the Interquartile Range (IQR) method.
df_clean = df.copy()

df_clean["Salary"] = df_clean["Salary"].fillna(
    df_clean["Salary"].median()
)

Q1 = df_clean["Salary"].quantile(0.25)
Q3 = df_clean["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean["Salary"] < lower_limit) |
    (df_clean["Salary"] > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:")
print(outliers)
```

```python
# Q12. Detect outliers using the Z-score method.
from scipy.stats import zscore

df_zscore = df.copy()

df_zscore["Salary"] = df_zscore["Salary"].fillna(
    df_zscore["Salary"].median()
)

df_zscore["Z_Score"] = zscore(df_zscore["Salary"])

outliers = df_zscore[
    abs(df_zscore["Z_Score"]) > 3
]

print(outliers)
```

```python
# Q13. Visualize outliers using a Box Plot.
import seaborn as sns
import matplotlib.pyplot as plt

df_clean["Salary"] = df_clean["Salary"].fillna(
    df_clean["Salary"].median()
)

sns.boxplot(y=df_clean["Salary"])

plt.title("Salary Box Plot")
plt.ylabel("Salary")

plt.show() 
```

```python
# Q14. Visualize outliers using a Scatter Plot.
sns.scatterplot(
    x=df_clean["Experience"],
    y=df_clean["Salary"]
)

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.show()
```

```python
# Q15. Remove outliers using the IQR method.
Q1 = df_clean["Salary"].quantile(0.25)
Q3 = df_clean["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df_iqr = df_clean[
    (df_clean["Salary"] >= lower_limit) &
    (df_clean["Salary"] <= upper_limit)
]

print(df_iqr)
```

```python
# Q16. Replace outliers with the median value.
df_median_outlier = df_clean.copy()

Q1 = df_median_outlier["Salary"].quantile(0.25)
Q3 = df_median_outlier["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

median_salary = df_median_outlier["Salary"].median()

outlier_condition = (
    (df_median_outlier["Salary"] < lower_limit) |
    (df_median_outlier["Salary"] > upper_limit)
)

df_median_outlier.loc[
    outlier_condition, "Salary"
] = median_salary

print(df_median_outlier)
```

```python
# Q17. Cap outliers using percentile-based capping.
df_capped = df_clean.copy()

lower = df_capped["Salary"].quantile(0.05)
upper = df_capped["Salary"].quantile(0.95)

df_capped["Salary"] = df_capped["Salary"].clip(
    lower=lower,
    upper=upper
)

print(df_capped)
```

```python
# Q18. Compare the dataset before and after outlier treatment.
comparison = pd.DataFrame({
    "Before": df_clean["Salary"],
    "After": df_capped["Salary"]
})

print(comparison)
```

```python
# Q33. Apply Log Transformation to skewed data.
import numpy as np
import matplotlib.pyplot as plt

df_log = df.copy()

df_log["Salary"] = df_log[
    "Salary"
].fillna(
    df_log["Salary"].median()
)

df_log["Log_Salary"] = np.log1p(
    df_log["Salary"]
)

print(df_log)

plt.hist(df_log["Log_Salary"], bins=10)

plt.title("Log Transformed Salary")
plt.xlabel("Log Salary")
plt.ylabel("Frequency")

plt.show()
```

```python
# Q34. Perform Feature Selection using correlation analysis.
import seaborn as sns
import matplotlib.pyplot as plt

df_corr = df.copy()

numeric_columns = [
    "Age",
    "Salary",
    "Experience"
]

for col in numeric_columns:
    df_corr[col] = df_corr[col].fillna(
        df_corr[col].median()
    )

correlation = df_corr[
    numeric_columns
].corr()

print(correlation)

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
# Q35. Create a final preprocessed dataset ready for Machine Learning.
import pandas as pd
from sklearn.preprocessing import StandardScaler

df_final = pd.read_csv("original_dataset.csv")

# Fill numerical missing values
numeric_columns = [
    "Age",
    "Salary",
    "Experience"
]

for col in numeric_columns:
    df_final[col] = df_final[col].fillna(
        df_final[col].median()
    )

# Fill categorical missing values
df_final["Department"] = df_final[
    "Department"
].fillna(
    df_final["Department"].mode()[0]
)

# One-hot encoding
df_final = pd.get_dummies(
    df_final,
    columns=["Department"],
    dtype=int
)

# Standard scaling
scaler = StandardScaler()

df_final[numeric_columns] = scaler.fit_transform(
    df_final[numeric_columns]
)

# Save final dataset
df_final.to_csv(
    "final_preprocessed_dataset.csv",
    index=False
)

print(df_final)
print("\nFinal preprocessed dataset saved successfully.")
```
